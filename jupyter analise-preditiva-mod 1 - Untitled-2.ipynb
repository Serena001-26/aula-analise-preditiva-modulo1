```python
# encontrar maior numero da lista
var_numeros = [0,1,2,3,4,5,6]
indice = 0
numero_max = var_numeros[0]
while indice < len(var_numeros):
    if var_numeros[indice] > numero_max:
        numero_max = var_numeros[indice]
    indice += 1
    print("o numero maximo e: " , numero_max)

```

```python
filmes = ["terror", "comedia", "drama", "acao", "aventura"]

# 1. Descobrimos o tamanho total da lista
quantidade = len(filmes)
print("Total de filmes na lista:", quantidade)

# 2. Pegamos o último item usando o tamanho - 1
# (Se tem 5 itens, o último índice é 5 - 1 = 4)
indice_do_ultimo = quantidade - 1
ultimo_filme = filmes[indice_do_ultimo]

print("O último filme da lista é:", ultimo_filme)
```
